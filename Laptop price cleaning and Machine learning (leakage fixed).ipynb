## Step 2: Cleaning Numerical Features

```python
# Inches should clearly be numerical data, it is saved an object. Let's examine unique values and clean invalid ones
laptop_raw.dtypes
display(laptop_raw['Inches'].unique())
# We can see there are values that are nan and there is one that is '?', let's replace them with np.nan
laptop_raw.replace(to_replace='?', value=np.nan, inplace=True)
laptop_raw.replace(to_replace='nan', value=np.nan, inplace=True)
laptop_raw['Inches'] = laptop_raw['Inches'].astype(np.float64)
display(laptop_raw.dtypes)
```

```python
display(laptop_raw['Memory'].unique())
# extract_storage_type looks for anything that has numbers in it then returns the alphabetical parts, giving us the storage type
def extract_storage_type(s):
    if pd.isnull(s):
        return np.nan
    parts = re.findall(r'\d+\.?\d*\s*(?:GB|TB)\s*([A-Za-z ]+)', s)
    return ' + '.join(part.strip() for part in parts)
# applying extract_storage type
laptop_raw['Storage Type'] = laptop_raw['Memory'].apply(extract_storage_type)

# This function totals our storage in GB, handles TB by converting to GB
def total_storage_in_gb(s):
    if not isinstance(s, str):
        return np.nan

    matches = re.findall(r'\d+\.?\d*\s*(?:GB|TB)', s)
    total_gb = 0

    for item in matches:
        number = float(re.search(r'\d+\.?\d*', item).group())
        if 'TB' in item:
            number *= 1024
        total_gb += number

    return round(float(total_gb), 0)
# applying total_storage_in_gb
laptop_raw['Memory'] = laptop_raw['Memory'].apply(total_storage_in_gb).astype('float64')
display(laptop_raw['Storage Type'].unique())
display(laptop_raw['Memory'].unique())
display(laptop_raw.dtypes)
```

```python
# Ram is another numerical feature saved as object, let's examine the unique values to confirm they are all in the same unit then strip non numeric characters
display(laptop_raw['Ram'].unique())
# Ram is all in GB, we will use this lambda function and some regex to strip away anything non numeric
laptop_raw['Ram'] = laptop_raw['Ram'].apply(lambda x: re.sub('[^0-9.]', '', x) if isinstance(x, str) else x).replace('', np.nan).astype(('float64'))
display(laptop_raw['Ram'].unique())
display(laptop_raw.dtypes)
```

```python
display(laptop_raw['Weight'].unique())
laptop_raw['Weight'] = laptop_raw['Weight'].apply(lambda x: re.sub('[^0-9.]', '', x) if isinstance(x, str) else x).replace('', np.nan).astype(('float64'))
laptop_raw['Weight'].replace(0.0002, np.nan)
display(laptop_raw['Weight'].unique())
```

```python
# here we narrow the CPUs down to brand and series
CPU_PATTERNS = [
    (r'AMD A\d+[ -]Series', 'AMD A-Series'),
    (r'AMD E\d*[ -]Series', 'AMD E-Series'),
    (r'AMD FX(?:\s*\d+)?', 'AMD FX'),
    (r'AMD Ryzen(?:\s*\d+)?', 'AMD Ryzen'),
    (r'Intel Atom(?:\s*\w+)?', 'Intel Atom'),
    (r'Intel Celeron Dual Core.*', 'Intel Celeron Dual Core'),
    (r'Intel Celeron Quad Core.*', 'Intel Celeron Quad Core'),
    (r'Intel Core i3(?:[- ]?\d{4,5}[A-Z]*)?', 'Intel Core i3'),
    (r'Intel Core i5(?:[- ]?\d{4,5}[A-Z]*)?', 'Intel Core i5'),
    (r'Intel Core i7(?:[- ]?\d{4,5}[A-Z]*)?', 'Intel Core i7'),
    (r'Intel Core M(?:\s*\w+)?', 'Intel Core M'),
    (r'Intel Pentium Dual Core.*', 'Intel Pentium Dual Core'),
    (r'Intel Pentium Quad Core.*', 'Intel Pentium Quad Core'),
    (r'Intel Xeon.*', 'Intel Xeon'),
]

def extract_cpus(df, column='Cpu', new_column='Cpu Brand'):
    def label_cpu(text):
        if pd.isnull(text):
            return ''
        for pat, label in CPU_PATTERNS:
            if re.search(pat, text, flags=re.IGNORECASE):
                return label
        return ''
    df[new_column] = df[column].apply(label_cpu)
    return df

laptop_raw = extract_cpus(laptop_raw, column='Cpu', new_column='Cpu Brand')

#The CPU pattern before produced over a dozen bars when visualized, to make this simpler we'll lump some Series together

CPU_PATTERNS = [
    (r'AMD A\d+[ -]Series', 'AMD Series'),
    (r'AMD E\d*[ -]Series', 'AMD Series'),
    (r'AMD Series', 'AMD Series'),
    (r'AMD FX(?:\s*\d+)?', 'AMD FX'),
    (r'AMD Ryzen(?:\s*\d+)?', 'AMD Ryzen'),
    (r'Intel Atom(?:\s*\w+)?', 'Intel Atom'),
    (r'Intel Celeron(?:\s+\w+)?', 'Intel Celeron'),
    (r'Intel Core i[3579](?:[- ]?\d{4,5}[A-Z]*)?', 'Intel Core'),
    (r'Intel Core M(?:\s*\w+)?', 'Intel Core'),
    (r'Intel Pentium(?:\s+\w+)?', 'Intel Pentium'),
    (r'Intel Xeon(?:.*)?', 'Intel Xeon')
]

laptop_raw = extract_cpus(laptop_raw, column='Cpu', new_column='Cpu Brand Concise')

display(laptop_raw.head())
display(laptop_raw['Cpu Brand Concise'].unique())
```

```python
display(laptop_raw['ScreenResolution'].unique())
def parse_screen_resolution(resolution_str):
    if pd.isnull(resolution_str):
        return None, None
    
    # Extract pixel resolution
    match = re.search(r'(\d+)\s*[xX]\s*(\d+)', resolution_str)
    pixel_count = None
    if match:
        width, height = map(int, match.groups())
        pixel_count = width * height
    
    # Detect marketing terms
    marketing_terms = []
    terms_map = {
        'Retina': r'\bRetina\b',
        '4K': r'\b4K\b|\b3840x2160\b',
        'UHD': r'\bUHD\b',
        'Full HD': r'\bFull\s*HD\b|\b1920x1080\b',
        'Touchscreen': r'\bTouch\b|\bTouchscreen\b',
        'IPS Panel': r'\bIPS\b',
        'OLED': r'\bOLED\b',
        'QHD': r'\bQHD\b|\b2560x1440\b'
    }
    
    for label, pattern in terms_map.items():
        if re.search(pattern, resolution_str, flags=re.IGNORECASE):
            marketing_terms.append(label)
    
    marketing_str = ', '.join(marketing_terms) if marketing_terms else None
    
    return pixel_count, marketing_str

# Applying parse_screen_resolution via a lambda and .apply
laptop_raw[['PixelCount', 'DisplayFeatures']] = laptop_raw['ScreenResolution'].apply(
    lambda x: pd.Series(parse_screen_resolution(x))
)
display(laptop_raw['PixelCount'].unique())
display(laptop_raw['DisplayFeatures'].unique())
display(laptop_raw.head())
```

## Step 6: Creating XGBoost Model

```python
import numpy as np
import pandas as pd
import xgboost as xgb

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

def train_xgb_pipeline(
    df,
    cat_cols,                 
    custom_orders,            
    num_cols,                 
    target_col="Price",
    log_target=True,
    random_state=42,
    test_size=0.2,
    n_estimators=1000
):
    # Drop rows with missing target 
    df_ = df.dropna(subset=[target_col]).copy()

    # Split
    X = df_.drop(columns=[target_col])
    y = df_[target_col].values
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state
    )

    # Build orders_aug to include imputer fill-values
    orders_aug = {}
    for col in cat_cols:
        base = list(custom_orders[col])
        missing_token = "None" if col == "DisplayFeatures" else "Unknown"
        if missing_token not in base:
            base = [missing_token] + base
        orders_aug[col] = base

    # Two categorical pipes to support different fill-values if desired
    cat_cols_disp = [c for c in cat_cols if c == "DisplayFeatures"]
    cat_cols_other = [c for c in cat_cols if c != "DisplayFeatures"]

    cat_pipe_other = Pipeline(steps=[
        ("impute", SimpleImputer(strategy="constant", fill_value="Unknown")),
        ("encode", OrdinalEncoder(
            categories=[orders_aug[c] for c in cat_cols_other],
            handle_unknown="use_encoded_value",
            unknown_value=-1,
            encoded_missing_value=-1
        ))
    ])

    cat_pipe_disp = None
    if cat_cols_disp:
        cat_pipe_disp = Pipeline(steps=[
            ("impute", SimpleImputer(strategy="constant", fill_value="None")),
            ("encode", OrdinalEncoder(
                categories=[orders_aug[c] for c in cat_cols_disp],
                handle_unknown="use_encoded_value",
                unknown_value=-1,
                encoded_missing_value=-1
            ))
        ])

    num_pipe = Pipeline(steps=[
        ("impute", SimpleImputer(strategy="median"))
    ])

    transformers = [
        ("num", num_pipe, num_cols),
        ("cat_other", cat_pipe_other, cat_cols_other),
    ]
    if cat_cols_disp:
        transformers.append(("cat_disp", cat_pipe_disp, cat_cols_disp))

    pre = ColumnTransformer(transformers=transformers, remainder="drop")

    reg = xgb.XGBRegressor(
        objective="reg:squarederror",
        n_estimators=n_estimators,
        learning_rate=0.03,
        max_depth=8,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        random_state=random_state,
        n_jobs=-1,
        tree_method="hist",
        verbosity=0
    )

    pipe = Pipeline([("pre", pre), ("reg", reg)])

    model = (TransformedTargetRegressor(regressor=pipe, func=np.log1p, inverse_func=np.expm1)
             if log_target else pipe)

    # Fit & evaluate
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    rmse = float(np.sqrt(mean_squared_error(y_test, y_pred)))
    mae  = float(mean_absolute_error(y_test, y_pred))
    r2   = float(r2_score(y_test, y_pred))
    n, p = X_test.shape
    adj_r2 = float(1 - (1 - r2) * (n - 1) / max(n - p - 1, 1))

    return {
        "model": model,
        "rmse_test": rmse,
        "mae_test": mae,
        "r2_test": r2,
        "adj_r2_test": adj_r2,
        "X_train_columns": X_train.columns.tolist(),
        "X_train": X_train,
        "X_test": X_test,
        "y_test": y_test,
}

cat_features = [
    'Company', 'TypeName', 'OpSys', 'Storage Type',
    'Cpu Brand Concise', 'Gpu Brand Concise', 'DisplayFeatures'
]
num_features = ['Memory', 'Inches', 'Weight', 'Ram', 'PixelCount']  

res = train_xgb_pipeline(
    df=df,
    cat_cols=cat_features,
    custom_orders=custom_orders,  
    num_cols=num_features,
    target_col="Price",
    log_target=True,
    random_state=42,
    test_size=0.2,
    n_estimators=1000
)

print(f"✅ Test RMSE: {res['rmse_test']:.2f}")
print(f"✅ Test MAE:  {res['mae_test']:.2f}")
print(f"✅ R²:        {res['r2_test']:.3f}")
print(f"✅ Adj R²:    {res['adj_r2_test']:.3f}")
```

## Step 7: Interpreting Shapley Values for AI Explainability

```python
# ---- SHAP cell that uses the splits returned in `res` ----
import numpy as np
import shap
from sklearn.compose import ColumnTransformer
import xgboost as xgb
from xgboost.sklearn import XGBRegressor as _XGBRegressor

# 0) Sanity: see what we have
print("[diag] res keys:", list(res.keys()))

# 1) Get the fitted pipeline (may be wrapped in TransformedTargetRegressor)
fitted = res["model"]
is_ttr = hasattr(fitted, "regressor_")
pipe = fitted.regressor_ if is_ttr else fitted

# 2) Get the preprocessor and trained regressor from the pipeline
try:
    preproc = pipe.named_steps["pre"]
except KeyError:
    # fallback by type
    preproc = next((s for s in pipe.named_steps.values() if isinstance(s, ColumnTransformer)), None)
    if preproc is None:
        raise RuntimeError("Could not find ColumnTransformer in pipeline.")

try:
    xgb_reg = pipe.named_steps["reg"]
except KeyError:
    xgb_reg = next((s for s in pipe.named_steps.values() if isinstance(s, _XGBRegressor)), None)
    if xgb_reg is None:
        raise RuntimeError("Could not find XGBRegressor in pipeline.")

print(f"[diag] preproc={type(preproc).__name__}, reg={type(xgb_reg).__name__}")

# 3) Use the returned X_test (features only!). Do NOT drop 'Price' here.
X_test = res["X_test"].copy()
print("[diag] X_test shape:", X_test.shape)

# 4) Transform with the same preprocessor
X_enc = preproc.transform(X_test)
print("[diag] X_enc shape:", getattr(X_enc, "shape", None))

# 5) (Optional) downsample for speed
max_rows = 3000
if X_enc.shape[0] > max_rows:
    rng = np.random.default_rng(42)
    idx = rng.choice(X_enc.shape[0], size=max_rows, replace=False)
    X_enc_s = X_enc[idx]
else:
    X_enc_s = X_enc

# 6) Feature names (robust)
try:
    feature_names = preproc.get_feature_names_out()
except Exception:
    names = []
    for name, transformer, cols in preproc.transformers_:
        if name == "remainder":
            continue
        if isinstance(cols, (list, tuple, np.ndarray)):
            names.extend(cols)
        else:
            names.append(cols)
    feature_names = np.array(names)

# 7) SHAP explainer (auto backend)
explainer = shap.Explainer(xgb_reg, X_enc_s)
sv = explainer(X_enc_s)

# 8) Handle SHAP API variants
shap_values = getattr(sv, "values", sv)
base_value = getattr(sv, "base_values", explainer.expected_value)
if isinstance(base_value, (list, np.ndarray)) and np.ndim(base_value) > 0:
    base_value = float(np.array(base_value).mean())

print(f"[diag] base_value={base_value:.4f}, shap_values shape={np.array(shap_values).shape}")
print("Note: if you trained with log_target=True, SHAP explains log(Price+1).")

# Example plot (uncomment one)
# shap.plots.beeswarm(sv, show=True)
shap.summary_plot(shap_values, X_enc_s, feature_names=feature_names)
```

```python
# ==== K-Fold on df + df2 (same schema), ORIGINAL COLUMNS ONLY, custom-order encoding + SHAP ====

import numpy as np
import pandas as pd
import shap
from sklearn.model_selection import KFold
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import xgboost as xgb

# ---------------- 0) Define which original columns to model ----------------
target_col = "Price"

# Original categorical + numeric columns (no engineered features)
cat_features_master = [
    'Company', 'TypeName', 'OpSys', 'Storage Type',
    'Cpu Brand Concise', 'Gpu Brand Concise', 'DisplayFeatures'
]
num_features_master = ['Memory', 'Inches', 'Weight', 'Ram', 'PixelCount']

# Predictors = all originals present in the data
# (filter to avoid key errors if any column is missing in either df)
all_predictors = []
for c in num_features_master + cat_features_master:
    if c in df.columns and c in df2.columns:
        all_predictors.append(c)

if not all_predictors:
    raise ValueError("No overlapping original predictor columns between df and df2.")

# Split into nums/cats actually present
cat_cols = [c for c in all_predictors if c in cat_features_master]
num_cols = [c for c in all_predictors if c in num_features_master]

# ---------------- 1) Stack df + df2 (Option A: same schema) ----------------
df1 = df.copy();   df1["__source__"] = "df1"
df2_ = df2.copy(); df2_["__source__"] = "df2"

common_cols = sorted(set(df1.columns) & set(df2_.columns))
df1 = df1[common_cols]
df2_ = df2_[common_cols]

df_all = pd.concat([df1, df2_], ignore_index=True)

# Keep only rows with target (safe pre-split)
df_all = df_all.dropna(subset=[target_col]).copy()

# Subset to predictors + target
missing = [c for c in all_predictors if c not in df_all.columns]
if missing:
    raise KeyError(f"Missing predictors in combined df: {missing}")

X_all = df_all[all_predictors].copy()
y_all = df_all[target_col].values

# ---------------- 2) Build custom-order encoder (augmented with fill tokens) ----------------
# Assumes you already defined `custom_orders` for categorical features
orders_aug = {}
for col in cat_cols:
    base = list(custom_orders[col])
    fill = "None" if col == "DisplayFeatures" else "Unknown"
    if fill not in base:
        base = [fill] + base
    orders_aug[col] = base

cat_cols_disp  = [c for c in cat_cols if c == "DisplayFeatures"]
cat_cols_other = [c for c in cat_cols if c != "DisplayFeatures"]

cat_pipe_other = Pipeline([
    ("impute", SimpleImputer(strategy="constant", fill_value="Unknown")),
    ("enc", OrdinalEncoder(
        categories=[orders_aug[c] for c in cat_cols_other] if cat_cols_other else [],
        handle_unknown="use_encoded_value",
        unknown_value=-1,
        encoded_missing_value=-1
    ))
])

cat_pipe_disp = None
if cat_cols_disp:
    cat_pipe_disp = Pipeline([
        ("impute", SimpleImputer(strategy="constant", fill_value="None")),
        ("enc", OrdinalEncoder(
            categories=[orders_aug[c] for c in cat_cols_disp],
            handle_unknown="use_encoded_value",
            unknown_value=-1,
            encoded_missing_value=-1
        ))
    ])

num_pipe = Pipeline([
    ("impute", SimpleImputer(strategy="median"))
])

transformers = []
if num_cols:       transformers.append(("num",       num_pipe,       num_cols))
if cat_cols_other: transformers.append(("cat_other", cat_pipe_other, cat_cols_other))
if cat_cols_disp:  transformers.append(("cat_disp",  cat_pipe_disp,  cat_cols_disp))

pre = ColumnTransformer(transformers, remainder="drop")

reg = xgb.XGBRegressor(
    objective="reg:squarederror",
    n_estimators=1000,
    learning_rate=0.03,
    max_depth=8,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    random_state=42,
    n_jobs=-1,
    tree_method="hist",
    verbosity=0
)

# If you’ve been using a log target, keep the wrapper; otherwise you can use model_pipe directly.
model_pipe = Pipeline([("pre", pre), ("reg", reg)])
model = TransformedTargetRegressor(regressor=model_pipe, func=np.log1p, inverse_func=np.expm1)

# ---------------- 3) K-Fold loop with per-fold metrics + SHAP collection ----------------
kf = KFold(n_splits=5, shuffle=True, random_state=42)

r2s, maes, rmses = [], [], []
X_shap_list = []
shap_vals_list = []

for fold, (tr_idx, va_idx) in enumerate(kf.split(X_all), 1):
    X_tr, X_va = X_all.iloc[tr_idx].copy(), X_all.iloc[va_idx].copy()
    y_tr, y_va = y_all[tr_idx], y_all[va_idx]

    # Fit (preprocessing refit per fold → leak-proof)
    model.fit(X_tr, y_tr)

    # Predict + metrics
    y_pred = model.predict(X_va)
    r2  = r2_score(y_va, y_pred)
    mae = mean_absolute_error(y_va, y_pred)
    rmse = np.sqrt(mean_squared_error(y_va, y_pred))
    r2s.append(r2); maes.append(mae); rmses.append(rmse)
    print(f"Fold {fold}: R²={r2:.3f}  MAE={mae:.2f}  RMSE={rmse:.2f}")

    # ----- SHAP on held-out fold -----
    # Unwrap to fitted pipeline for THIS fold
    fitted_pipe = model.regressor_ if hasattr(model, "regressor_") else model
    fitted_pre  = fitted_pipe.named_steps["pre"]
    fitted_reg  = fitted_pipe.named_steps["reg"]

    # Transform validation features
    X_va_enc = fitted_pre.transform(X_va)

    # Feature names (strip transformer prefixes for readability)
    try:
        feat_names = fitted_pre.get_feature_names_out()
        feat_names = [n.split("__", 1)[-1] for n in feat_names]  # remove 'num__', 'cat_other__', etc.
    except Exception:
        # Fallback: follow ColumnTransformer order
        names = []
        for name, transformer, cols in fitted_pre.transformers_:
            if name == "remainder":
                continue
            names.extend(list(cols) if isinstance(cols, (list, tuple, np.ndarray)) else [cols])
        feat_names = names

    # Sample for speed
    n_sample = min(600, X_va_enc.shape[0])
    rng = np.random.default_rng(42 + fold)
    idx = rng.choice(X_va_enc.shape[0], size=n_sample, replace=False)
    Xv_enc = X_va_enc[idx]

    # SHAP explainer
    explainer = shap.Explainer(fitted_reg, Xv_enc)
    sv = explainer(Xv_enc)  # Explanation object

    # Store for combined summary
    X_shap_list.append(pd.DataFrame(Xv_enc, columns=feat_names))
    shap_vals_list.append(sv)

# ---------------- 4) Aggregated results ----------------
print("\n=== 5-fold CV (original columns only) ===")
print(f"R²   mean={np.mean(r2s):.3f} ± {np.std(r2s):.3f}")
print(f"MAE  mean={np.mean(maes):.2f}")
print(f"RMSE mean={np.mean(rmses):.2f}")

# ---------------- 5) Combined SHAP summary across folds ----------------
X_shap_all = pd.concat(X_shap_list, axis=0)

# Concatenate Explanation objects
shap_all = shap.Explanation(
    values=np.vstack([sv.values for sv in shap_vals_list]),
    base_values=np.concatenate([np.atleast_1d(sv.base_values) for sv in shap_vals_list]),
    data=np.vstack([sv.data for sv in shap_vals_list]),
    feature_names=list(X_shap_all.columns)
)

# Plot (uncomment your preferred API)
# shap.plots.beeswarm(shap_all, show=True)                # modern
shap.summary_plot(shap_all.values, X_shap_all, feature_names=shap_all.feature_names)
```
